```python
def c_bmi(h,w):
  bmi=w/(h/100)**2
  return bmi
bmi1=(170,60)
bmi2=(160,55)
bmi3=(160,80)
print(f"第一位bmi{c_bmi(*bmi1):.2f}\n第二位bmi{c_bmi(*bmi2):.2f}\n第三位bmi{c_bmi(*bmi3):.2f}")
```
