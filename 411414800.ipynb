```python
def add(p, q, dis=0.05):
    total = p * q
    tax = total * dis
    total_all = total + tax
    return total, tax, total_all

try:
    p = int(input("請輸入單價："))
    q = int(input("請輸入數量："))
    x, y, z = add(p, q)
    print(f"小計：{x}")
    print(f"稅金：{y}")
    print(f"總金額：{z}")
except ValueError:
    print("請輸入正確的數值！")
```
