```python
listA = []
listB = []

# 輸入 A 倉文具
while True:
    item = input("請輸入A倉文具, 輸入 '/' 結束：")
    if item == "/":
        break
    listA.append(item)

# 輸入 B 倉文具
while True:
    item = input("請輸入B倉文具, 輸入 '/' 結束：")
    if item == "/":
        break
    listB.append(item)

# 合併清單並加入膠水
listC = listA + listB
listC.append("膠水")

# 使用 '/' 連結清單元素
result = "/".join(listC)
print(f"二倉合併清單：{result}")
```
